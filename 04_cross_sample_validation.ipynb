```python
from pathlib import Path
import os, json
import numpy as np
import pandas as pd
from scipy import stats
from scipy.optimize import linear_sum_assignment
from scipy.linalg import orthogonal_procrustes
from threadpoolctl import threadpool_limits
PCS = [f'PC{i}' for i in range(1, 6)]

def bh_adjust(p):
    p = np.asarray(p, dtype=float)
    order = np.argsort(p)
    q = np.minimum.accumulate((p[order] * len(p) / np.arange(1, len(p) + 1))[::-1])[::-1]
    result = np.empty_like(q)
    result[order] = np.minimum(q, 1)
    return result

def congruence_matrix(a, b):
    return a.T @ b / np.outer(np.linalg.norm(a, axis=0), np.linalg.norm(b, axis=0))

def pca(z):
    with threadpool_limits(limits=1):
        u, s, vt = np.linalg.svd(z, full_matrices=False)
    eigen = s * s / (len(z) - 1)
    vectors = vt.T
    scores = z @ vectors
    loadings = vectors * np.sqrt(eigen)
    anchors = np.abs(loadings).argmax(axis=0)
    signs = np.sign(loadings[anchors, np.arange(loadings.shape[1])])
    signs[signs == 0] = 1
    return {
        'eigenvalues': eigen,
        'vectors': vectors * signs,
        'scores': scores * signs,
        'loadings': loadings * signs,
        'anchors': anchors,
        'signs': signs
    }

def parallel_analysis(z, n_iter, seed):
    rng = np.random.default_rng(seed)
    draws = np.empty((n_iter, z.shape[1]))
    with threadpool_limits(limits=1):
        for k in range(n_iter):
            perm = np.column_stack([rng.permutation(z[:, j]) for j in range(z.shape[1])])
            draws[k] = np.linalg.eigvalsh(perm.T @ perm / (len(z) - 1))[::-1]
    return draws

def frame_matrix(a, rows, cols, index_name):
    d = pd.DataFrame(a, index=rows, columns=cols)
    d.index.name = index_name
    return d.reset_index()

def correlation_rows(lp, lv, sp, sv, method):
    rows = []
    for j, pc in enumerate(PCS):
        lr = stats.pearsonr(lp[:, j], lv[:, j])
        pr = stats.pearsonr(sp[:, j], sv[:, j])
        sr = stats.spearmanr(sp[:, j], sv[:, j])
        rows.append(
            {
                'Method': method,
                'Primary component': pc,
                'Loading congruence': float(congruence_matrix(lp, lv)[j, j]),
                'Loading Pearson r': lr.statistic,
                'Loading Pearson p': lr.pvalue,
                'Score Pearson r': pr.statistic,
                'Score Pearson p': pr.pvalue,
                'Score Spearman rho': sr.statistic,
                'Score Spearman p': sr.pvalue,
                'Loading RMSE': np.sqrt(np.mean((lp[:, j] - lv[:, j]) ** 2)),
                'Standardized-score RMSE': np.sqrt(np.mean((sp[:, j] - sv[:, j]) ** 2)),
                'Word-score sign agreement': np.mean(np.sign(sp[:, j]) == np.sign(sv[:, j]))
            }
        )
    df = pd.DataFrame(rows)
    for stem in ['Loading Pearson', 'Score Pearson', 'Score Spearman']:
        df[stem + ' BH q'] = bh_adjust(df[stem + ' p'])
    return df

def export_workbook_local(tables, path):
    with pd.ExcelWriter(path, engine='openpyxl') as writer:
        for name, table in tables.items():
            table.to_excel(writer, sheet_name=name, index=False)

def make_figures(out, pa, pca_p, pca_v, sp, sv, groups):
    import matplotlib
    matplotlib.use('Agg')
    import matplotlib.pyplot as plt
    from matplotlib import font_manager as fm
    font = Path(os.environ.get('ARIAL_FONT_PATH', 'fonts/arial.ttf'))
    if font.is_file():
        fm.fontManager.addfont(str(font))
        plt.rcParams['font.family'] = fm.FontProperties(fname=font).get_name()
    plt.rcParams.update({'font.size': 8, 'pdf.fonttype': 42, 'svg.fonttype': 'none'})
    figure_dir = out / '05_figures'
    figure_dir.mkdir(exist_ok=True)
    fig, axs = plt.subplots(1, 2, figsize=(9, 3.3))
    for ax, sample, pp in zip(axs, ['Primary', 'Validation'], [pca_p, pca_v]):
        d = pa[pa.Sample.eq(sample)].iloc[:12]
        ax.plot(d['Rank'], d['Observed eigenvalue'], 'o-', label='Observed', ms=3)
        ax.plot(d['Rank'], d['Null 95th percentile'], '--', label='Permutation 95th percentile')
        ax.set(title=sample, xlabel='Component rank', ylabel='Eigenvalue')
        ax.legend(fontsize=7)
        ax.spines[['top', 'right']].set_visible(False)
    fig.tight_layout()
    fig.savefig(figure_dir / '01_parallel_analysis.png', dpi=180)
    fig.savefig(figure_dir / '01_parallel_analysis.pdf')
    plt.close(fig)
    fig, axs = plt.subplots(1, 5, figsize=(12, 2.5))
    for j, ax in enumerate(axs):
        ax.scatter(sp[:, j], sv[:, j], s=10, alpha=0.7, color='#2066AA')
        ax.plot([-3, 3], [-3, 3], color='#AAAAAA', lw=0.7)
        ax.set(xlim=(-3, 3), ylim=(-3, 3), aspect='equal', title=PCS[j], xlabel='Primary score')
        if j == 0:
            ax.set_ylabel('Validation score')
        ax.spines[['top', 'right']].set_visible(False)
    fig.tight_layout()
    fig.savefig(figure_dir / '02_independent_axis_scores.png', dpi=180)
    fig.savefig(figure_dir / '02_independent_axis_scores.pdf')
    plt.close(fig)
    colors = ['#BA3C48', '#8C6BB1', '#C28B38', '#368C89', '#2066AA']
    fig, axs = plt.subplots(1, 5, figsize=(12, 9), sharey=True)
    locals_ = groups['Local semantic group'].cat.categories.tolist()
    broad = groups['Broad semantic group'].cat.categories.tolist()
    for j, ax in enumerate(axs):
        for y, local in enumerate(locals_):
            idx = np.flatnonzero(groups['Local semantic group'].eq(local))
            c = colors[broad.index(groups.iloc[idx[0]]['Broad semantic group'])]
            ax.scatter(
                sp[idx, j],
                np.full(len(idx), y - 0.1),
                s=13,
                facecolors='none',
                edgecolors=c,
                lw=0.8
            )
            ax.scatter(sv[idx, j], np.full(len(idx), y + 0.1), s=12, color=c, alpha=0.8)
        ax.axvline(0, color='#B0B0B0', lw=0.6)
        ax.set(xlim=(-3, 3), title=PCS[j], xlabel='Standardized score')
        ax.spines[['top', 'right', 'left']].set_visible(False)
    axs[0].set_yticks(range(15), locals_, fontsize=7)
    axs[0].invert_yaxis()
    fig.suptitle('Independent axes: primary (open) and validation (filled)', fontsize=10)
    fig.tight_layout()
    fig.savefig(figure_dir / '03_group_distributions.png', dpi=180)
    fig.savefig(figure_dir / '03_group_distributions.pdf')
    plt.close(fig)

def run_analysis(
    primary_path,
    validation_path,
    reference_path,
    group_path,
    out_dir,
    n_permutations=N_PERMUTATIONS,
    seed=RANDOM_SEED,
    export_xlsx=True
):
    out = Path(out_dir)
    out.mkdir(parents=True, exist_ok=True)
    folders = {
        name: out / name for name in [
            '00_inputs',
            '01_aggregation',
            '02_independent_pca',
            '03_correspondence',
            '04_group_distributions',
            '05_figures',
            '06_report'
        ]
    }
    for d in folders.values():
        d.mkdir(exist_ok=True)
    reference_path = Path(reference_path)
    primary_path = Path(primary_path)
    validation_path = Path(validation_path)
    reference_loadings = pd.read_excel(
        reference_path,
        sheet_name='Component_Loadings'
    ).set_index('Emotion Feature')[PCS]
    reference_scores = pd.read_excel(
        reference_path,
        sheet_name='Word_Scores'
    ).set_index('Emotion Word')[PCS]
    words = reference_scores.index.tolist()
    features = reference_loadings.index.tolist()
    assert len(words) == 70 and len(features) == 43
    semantic_source = pd.read_csv(group_path)[
        ['Emotion Word', 'Broad semantic group', 'Local semantic group']
    ]
    assert semantic_source['Emotion Word'].is_unique and set(semantic_source['Emotion Word']) == set(words)
    groups = semantic_source.set_index('Emotion Word').reindex(words).reset_index()
    for col in ['Broad semantic group', 'Local semantic group']:
        groups[col] = pd.Categorical(
            groups[col],
            categories=semantic_source[col].drop_duplicates().tolist(),
            ordered=True
        )
    semantic_source.to_csv(folders['00_inputs'] / 'Semantic_Group_Membership.csv', index=False)
    tables = {}
    qc = []
    aggregates = {}
    matrices = {}
    results = {}
    null_draws = {}
    pa_tables = []
    axis_orientation = []
    for sample, path, expected_n in [
        ('Primary', primary_path, 2372),
        ('Validation', validation_path, 304)
    ]:
        raw = pd.read_excel(path)
        required = ['sub_id', 'group', 'word_raw_text', 'dim_name', 'rating']
        assert all((c in raw for c in required))
        missing = int(raw[required].isna().any(axis=1).sum())
        duplicate = int(raw.duplicated(['sub_id', 'word_raw_text', 'dim_name']).sum())
        nonnumeric = pd.to_numeric(raw.rating, errors='coerce')
        invalid = int((~nonnumeric.between(1, 9) | nonnumeric.isna() | (nonnumeric % 1 != 0)).sum())
        assert missing == duplicate == invalid == 0, 'Input integrity failed; no records have been silently removed.'
        assert set(raw.word_raw_text) == set(words) and set(raw.dim_name) == set(features)
        assert raw.sub_id.nunique() == expected_n
        raw = raw.copy()
        raw.rating = nonnumeric
        ag = raw.groupby(['word_raw_text', 'dim_name'], sort=False).rating.agg(n='count', mean='mean', sd='std').reset_index().rename(
            columns={'word_raw_text': 'Emotion Word', 'dim_name': 'Emotion Feature'}
        )
        ag['se'] = ag.sd / np.sqrt(ag.n)
        ag.insert(0, 'Sample', sample)
        aggregates[sample] = ag
        assert len(ag) == 3010
        ag.to_csv(folders['01_aggregation'] / f'{sample}_Cell_Statistics.csv', index=False)
        m = ag.pivot(
            index='Emotion Word',
            columns='Emotion Feature',
            values='mean'
        ).reindex(index=words, columns=features)
        assert m.notna().all().all()
        z = (m - m.mean()) / m.std(ddof=1)
        matrices[sample] = m
        for label, data in [('Mean_Matrix', m), ('Standardized_Matrix', z)]:
            data.to_csv(folders['01_aggregation'] / f'{sample}_{label}.csv')
        pd.DataFrame(
            {
                'Emotion Feature': features,
                'Mean across words': m.mean().values,
                'Sample SD across words': m.std(ddof=1).values
            }
        ).to_csv(folders['01_aggregation'] / f'{sample}_Standardization.csv', index=False)
        res = pca(z.to_numpy())
        results[sample] = res
        for k in range(5):
            axis_orientation.append(
                {
                    'Sample': sample,
                    'Component': PCS[k],
                    'Anchor feature': features[res['anchors'][k]],
                    'Sign multiplier': res['signs'][k]
                }
            )
        for key, label, ids in [
            ('loadings', 'Loadings', features),
            ('vectors', 'Eigenvectors', features),
            ('scores', 'Scores', words)
        ]:
            frame_matrix(
                res[key],
                ids,
                [f'PC{i + 1}' for i in range(43)],
                'Emotion Feature' if key != 'scores' else 'Emotion Word'
            ).to_csv(folders['02_independent_pca'] / f'{sample}_{label}_All43.csv', index=False)
        null = parallel_analysis(z.to_numpy(), n_permutations, seed)
        null_draws[sample] = null
        np.savez_compressed(
            folders['02_independent_pca'] / f'{sample}_Permutation_Eigenvalues.npz',
            eigenvalues=null,
            seed=seed,
            n_permutations=n_permutations
        )
        pd.DataFrame(null, columns=[f'PC{i + 1}' for i in range(43)]).to_csv(
            folders['02_independent_pca'] / f'{sample}_Permutation_Eigenvalues.csv.gz',
            index_label='Iteration',
            compression='gzip'
        )
        threshold = np.percentile(null, 95, axis=0)
        above = res['eigenvalues'] > threshold
        retained = np.logical_and.accumulate(above)
        pa = pd.DataFrame(
            {
                'Sample': sample,
                'Rank': np.arange(1, 44),
                'Observed eigenvalue': res['eigenvalues'],
                'Variance proportion': res['eigenvalues'] / 43,
                'Cumulative variance proportion': np.cumsum(res['eigenvalues']) / 43,
                'Null mean': null.mean(axis=0),
                'Null 2.5th percentile': np.percentile(null, 2.5, axis=0),
                'Null 95th percentile': threshold,
                'Null 97.5th percentile': np.percentile(null, 97.5, axis=0),
                'Exceeds threshold': above,
                'Retained sequentially': retained
            }
        )
        pa_tables.append(pa)
        pa.to_csv(folders['02_independent_pca'] / f'{sample}_Parallel_Analysis.csv', index=False)
        qc.append(
            {
                'Sample': sample,
                'Participants': raw.sub_id.nunique(),
                'Rating records': len(raw),
                'Rating versions': raw.group.nunique(),
                'Emotion words': len(words),
                'Features': len(features),
                'Covered cells': len(ag),
                'Minimum ratings per cell': int(ag.n.min()),
                'Maximum ratings per cell': int(ag.n.max()),
                'Missing required values': missing,
                'Duplicate participant-word-feature records': duplicate,
                'Invalid ratings': invalid,
                'Retained components': int(retained.sum())
            }
        )
    primary, validation = (results['Primary'], results['Validation'])
    errL = np.abs(primary['loadings'][:, :5] - reference_loadings.values).max()
    errS = np.abs(primary['scores'][:, :5] - reference_scores.values).max()
    assert errL < 1e-08 and errS < 1e-08, 'Primary PCA did not reproduce supplied reference.'
    reference_check = pd.DataFrame(
        [
            {
                'Comparison': 'Primary loading matrix',
                'Maximum absolute difference': errL,
                'Tolerance': 1e-08,
                'Reproduced': errL < 1e-08
            },
            {
                'Comparison': 'Primary component scores',
                'Maximum absolute difference': errS,
                'Tolerance': 1e-08,
                'Reproduced': errS < 1e-08
            }
        ]
    )
    lp = primary['loadings'][:, :5]
    lv = validation['loadings'][:, :5]
    sp = primary['scores'][:, :5] / np.sqrt(primary['eigenvalues'][:5])
    sv = validation['scores'][:, :5] / np.sqrt(validation['eigenvalues'][:5])
    cross = congruence_matrix(lp, lv)
    row, order = linear_sum_assignment(-np.abs(cross))
    assert np.array_equal(row, np.arange(5))
    signs = np.sign(cross[row, order])
    signs[signs == 0] = 1
    lm = lv[:, order] * signs
    sm = sv[:, order] * signs
    mapping = pd.DataFrame(
        {
            'Primary component': PCS,
            'Validation original component': [PCS[x] for x in order],
            'Sign multiplier': signs,
            'Congruence before sign alignment': cross[row, order]
        }
    )
    main = correlation_rows(lp, lm, sp, sm, 'Independent axes: permutation and sign only')
    main = main.merge(mapping, on='Primary component', how='left')
    first = [
        'Primary component',
        'Validation original component',
        'Sign multiplier',
        'Loading congruence',
        'Loading Pearson r',
        'Score Pearson r',
        'Score Spearman rho'
    ]
    main = main[first + [c for c in main if c not in first]]
    R, _ = orthogonal_procrustes(lv, lp)
    lr = lv @ R
    sr = sv @ R
    supplementary = correlation_rows(lp, lr, sp, sr, 'Supplementary: orthogonal Procrustes')
    zp = pd.read_csv(
        folders['01_aggregation'] / 'Primary_Standardized_Matrix.csv',
        index_col=0
    ).to_numpy()
    zv = pd.read_csv(
        folders['01_aggregation'] / 'Validation_Standardized_Matrix.csv',
        index_col=0
    ).to_numpy()
    realized = np.corrcoef(zv.T, sr.T)[:43, 43:]
    assert np.max(np.abs(realized - lr)) < 1e-10
    diagnostics = pd.DataFrame(
        {
            'Aligned axis': PCS,
            'Whitened-score loading identity max error': np.max(abs(realized - lr), axis=0)
        }
    )
    for label, data, ids in [
        ('Primary_Loadings', lp, features),
        ('Validation_Matched_Loadings', lm, features),
        ('Validation_Procrustes_Loadings', lr, features),
        ('Primary_Standardized_Scores', sp, words),
        ('Validation_Matched_Standardized_Scores', sm, words),
        ('Validation_Procrustes_Standardized_Scores', sr, words)
    ]:
        frame_matrix(
            data,
            ids,
            PCS,
            'Emotion Word' if 'Scores' in label else 'Emotion Feature'
        ).to_csv(folders['03_correspondence'] / f'{label}.csv', index=False)
    pd.DataFrame(R, index=PCS, columns=PCS).to_csv(
        folders['03_correspondence'] / 'Procrustes_Rotation_Matrix.csv',
        index_label='Validation original component'
    )
    mapping.to_csv(folders['03_correspondence'] / 'Independent_Axis_Matching.csv', index=False)
    main.to_csv(folders['03_correspondence'] / 'Independent_Axis_Correspondence.csv', index=False)
    supplementary.to_csv(
        folders['03_correspondence'] / 'Procrustes_Correspondence.csv',
        index=False
    )
    diagnostics.to_csv(
        folders['03_correspondence'] / 'Procrustes_Score_Consistency.csv',
        index=False
    )
    all_scores = []
    all_loadings = []
    for j, pc in enumerate(PCS):
        for i, word in enumerate(words):
            all_scores.append(
                {
                    'Emotion Word': word,
                    'Component': pc,
                    'Broad semantic group': groups.iloc[i]['Broad semantic group'],
                    'Local semantic group': groups.iloc[i]['Local semantic group'],
                    'Primary standardized score': sp[i, j],
                    'Validation matched score': sm[i, j],
                    'Validation Procrustes score': sr[i, j]
                }
            )
        for i, feature in enumerate(features):
            all_loadings.append(
                {
                    'Emotion Feature': feature,
                    'Component': pc,
                    'Primary loading': lp[i, j],
                    'Validation matched loading': lm[i, j],
                    'Validation Procrustes loading': lr[i, j],
                    'Matched minus primary': lm[i, j] - lp[i, j]
                }
            )
    score_table = pd.DataFrame(all_scores)
    loading_table = pd.DataFrame(all_loadings)
    gd = []
    for method, valid_scores in [('Independent', sm), ('Procrustes', sr)]:
        for level, column in [('Broad', 'Broad semantic group'), ('Local', 'Local semantic group')]:
            for name, g in groups.groupby(column, sort=True, observed=True):
                idx = g.index.to_numpy()
                for j, pc in enumerate(PCS):
                    item = {
                        'Alignment': method,
                        'Level': level,
                        'Group': name,
                        'Component': pc,
                        'Number of words': len(idx)
                    }
                    for sample, values in [
                        ('Primary', sp[idx, j]),
                        ('Validation', valid_scores[idx, j])
                    ]:
                        item.update(
                            {
                                f'{sample} mean': float(np.mean(values)),
                                f'{sample} SD': float(np.std(values, ddof=1)),
                                f'{sample} median': float(np.median(values)),
                                f'{sample} Q1': float(np.percentile(values, 25)),
                                f'{sample} Q3': float(np.percentile(values, 75)),
                                f'{sample} minimum': float(min(values)),
                                f'{sample} maximum': float(max(values)),
                                f'{sample} positive count': int(sum(values > 0)),
                                f'{sample} negative count': int(sum(values < 0))
                            }
                        )
                    item['Word-score sign agreement'] = float(
                        np.mean(np.sign(sp[idx, j]) == np.sign(valid_scores[idx, j]))
                    )
                    gd.append(item)
    group_stats = pd.DataFrame(gd)
    local_contrasts = [
        (1, 'Nostalgia and homesickness', 'Sadness, distress, and loneliness'),
        (2, 'Positive anticipation', 'Compassion and caring'),
        (2, 'Love, happiness, and being moved', 'Compassion and caring'),
        (2, 'Confidence and pride', 'Compassion and caring'),
        (2, 'Fear and worry', 'Surprise, curiosity, and confusion'),
        (2, 'Dissatisfaction, aversion, and self-reproach', 'Shyness, embarrassment, and shame'),
        (3, 'Compassion and caring', 'Confidence and pride'),
        (3, 'Adoration and awe', 'Confidence and pride'),
        (4, 'Fear and worry', 'Surprise, curiosity, and confusion'),
        (4, 'Sadness, distress, and loneliness', 'Frustration and dejection'),
        (5, 'Aesthetic and pleasurable experiences', 'Calmness, comfort, and satisfaction'),
        (5, 'Confidence and pride', 'Positive anticipation'),
        (5, 'Confidence and pride', 'Adoration and awe'),
        (5, 'Shyness, embarrassment, and shame', 'Dissatisfaction, aversion, and self-reproach')
    ]
    contrasts = []
    feature_contrasts = []
    for pc_num, A, B in local_contrasts:
        j = pc_num - 1
        ia = groups.index[groups['Local semantic group'].eq(A)].to_numpy()
        ib = groups.index[groups['Local semantic group'].eq(B)].to_numpy()
        assert len(ia) > 0 and len(ib) > 0
        dp = float(sp[ia, j].mean() - sp[ib, j].mean())
        dv = float(sm[ia, j].mean() - sm[ib, j].mean())
        dr = float(sr[ia, j].mean() - sr[ib, j].mean())
        contrasts.append(
            {
                'Component': PCS[j],
                'Subgroup A': A,
                'Subgroup B': B,
                'A words': len(ia),
                'B words': len(ib),
                'Primary mean difference A-B': dp,
                'Validation independent difference A-B': dv,
                'Validation Procrustes difference A-B': dr,
                'Independent direction preserved': int(dp * dv > 0),
                'Procrustes direction preserved': int(dp * dr > 0)
            }
        )
        feature_idx = np.r_[np.argsort(-lp[:, j])[:4], np.argsort(lp[:, j])[:2]]
        for k in feature_idx:
            row = {
                'Component': PCS[j],
                'Subgroup A': A,
                'Subgroup B': B,
                'Emotion Feature': features[k],
                'Primary loading': lp[k, j]
            }
            for sample, m in matrices.items():
                av = float(m.iloc[ia, k].mean())
                bv = float(m.iloc[ib, k].mean())
                row.update(
                    {
                        f'{sample} A mean rating': av,
                        f'{sample} B mean rating': bv,
                        f'{sample} difference A-B': av - bv
                    }
                )
            row['Feature-difference direction preserved'] = int(
                row['Primary difference A-B'] * row['Validation difference A-B'] > 0
            )
            feature_contrasts.append(row)
    contrast_table = pd.DataFrame(contrasts)
    feature_contrast_table = pd.DataFrame(feature_contrasts)
    for name, df in [('Word_Score_Comparison', score_table), ('Loading_Comparison', loading_table)]:
        df.to_csv(folders['03_correspondence'] / f'{name}.csv', index=False)
    for name, df in [
        ('Group_Distributions', group_stats),
        ('Descriptive_Subgroup_Contrasts', contrast_table),
        ('Feature_Contrast_Details', feature_contrast_table)
    ]:
        df.to_csv(folders['04_group_distributions'] / f'{name}.csv', index=False)
    primary_pa = pa_tables[0]
    validation_pa = pa_tables[1]
    summary_rows = [
        ('Analysis', 'Cross-sample replication of individually interpreted PCA axes'),
        ('Primary participants', qc[0]['Participants']),
        ('Validation participants', qc[1]['Participants']),
        ('Primary valid ratings', qc[0]['Rating records']),
        ('Validation valid ratings', qc[1]['Rating records']),
        ('Primary retained components', int(primary_pa['Retained sequentially'].sum())),
        ('Validation retained components', int(validation_pa['Retained sequentially'].sum())),
        ('Primary first-five variance proportion', float(primary['eigenvalues'][:5].sum() / 43)),
        (
            'Validation first-five variance proportion',
            float(validation['eigenvalues'][:5].sum() / 43)
        ),
        ('Main loading congruence minimum', float(main['Loading congruence'].min())),
        ('Main loading congruence maximum', float(main['Loading congruence'].max())),
        ('Main score Pearson minimum', float(main['Score Pearson r'].min())),
        ('Main score Pearson maximum', float(main['Score Pearson r'].max())),
        (
            'Supplementary loading congruence minimum',
            float(supplementary['Loading congruence'].min())
        ),
        (
            'Supplementary loading congruence maximum',
            float(supplementary['Loading congruence'].max())
        ),
        ('Supplementary score Pearson minimum', float(supplementary['Score Pearson r'].min())),
        ('Supplementary score Pearson maximum', float(supplementary['Score Pearson r'].max())),
        ('Descriptive local contrasts', len(contrast_table)),
        (
            'Independent contrast directions preserved',
            int(contrast_table['Independent direction preserved'].sum())
        ),
        (
            'Procrustes contrast directions preserved',
            int(contrast_table['Procrustes direction preserved'].sum())
        ),
        ('Permutations per sample', n_permutations),
        ('Random seed per sample', seed)
    ]
    tables['00_Summary'] = pd.DataFrame(summary_rows, columns=['Item', 'Value'])
    tables['01_Axis_Correspondence'] = main
    tables['02_Subgroup_Contrasts'] = contrast_table
    tables['03_Sample_QC'] = pd.DataFrame(qc)
    tables['04_Reference_Check'] = reference_check
    tables['05_Parallel_Analysis'] = pd.concat(pa_tables, ignore_index=True)
    tables['06_Loading_Comparison'] = loading_table
    tables['07_Word_Scores'] = score_table
    tables['08_Group_Distributions'] = group_stats[group_stats.Alignment.eq('Independent')].reset_index(drop=True)
    tables['09_Feature_Contrasts'] = feature_contrast_table
    tables['10_Procrustes_Comparison'] = supplementary
    tables['11_Rotation_Matrix'] = frame_matrix(R, PCS, PCS, 'Validation original component')
    tables['12_Rotation_Diagnostics'] = diagnostics
    tables['13_Congruence_Crossmatrix'] = frame_matrix(cross, PCS, PCS, 'Primary component')
    tables['14_Sign_Orientation'] = pd.DataFrame(axis_orientation)
    for sample, m in matrices.items():
        tables[
            f'16_{sample}_Means' if sample == 'Primary' else '17_Validation_Means'
        ] = m.reset_index()
    tables['18_Cell_Statistics'] = pd.concat(aggregates.values(), ignore_index=True)
    tables['19_Semantic_Groups'] = groups
    tables['20_Procrustes_Groups'] = group_stats[group_stats.Alignment.eq('Procrustes')].reset_index(drop=True)
    payload = {
        name: {
            'columns': list(df.columns),
            'rows': df.astype(object).where(pd.notna(df), None).values.tolist()
        } for name,
        df in tables.items()
    }
    (folders['06_report'] / 'Workbook_Tables.json').write_text(
        json.dumps(payload, ensure_ascii=False, allow_nan=False, indent=1),
        encoding='utf-8'
    )
    make_figures(out, tables['05_Parallel_Analysis'], primary, validation, sp, sm, groups)
    np.savez_compressed(
        folders['03_correspondence'] / 'PCA_and_Alignment_Arrays.npz',
        primary_eigenvalues=primary['eigenvalues'],
        validation_eigenvalues=validation['eigenvalues'],
        primary_loadings=lp,
        validation_loadings=lv,
        matched_loadings=lm,
        procrustes_loadings=lr,
        primary_scores=sp,
        validation_scores=sv,
        matched_scores=sm,
        procrustes_scores=sr,
        rotation=R,
        matched_order=order,
        matched_signs=signs
    )
    if export_xlsx:
        export_workbook_local(tables, folders['06_report'] / 'cross_sample_validation_summary.xlsx')
    return tables
```
